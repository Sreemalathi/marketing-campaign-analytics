# Income Column Analysis

```python
# Convert the 'Dt_Customer' column to datetime format
df['Dt_Customer'] = pd.to_datetime(df['Dt_Customer'])
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df['Income'].describe()
```

```python
Q1 = 35303
Q3 = 68522
IQR = Q3 - Q1
IQR
```

```python
upper_limit = Q3 + 1.5 * IQR
lower_limit = Q1 - 1.5 * IQR
```

```python
(df['Income'] > upper_limit).sum()
# Number of rows where Income is greater than the upper limit
# only ID9432 seems to be an outlier with Income = 666666 too high compared to the rest of the data.
```

```python
(df['Income'] > lower_limit).sum()
```

```python
df.loc[df.isna().any(axis=1),:]
```

```python
# Fill missing values in the Income column with the median value of the column
median_income = df['Income'].median()
df['Income'] = df['Income'].fillna(median_income)
```

```python
# replace the outlier value of 666666 with the median value of the Income column
df.loc[df['Income'] == 666666, 'Income'] = median_income
```

```python
df['Income'].max()
df[df['ID'] == 9432][['ID', 'Income']]
```

# Year Column Analysis

```python
df['Year_Birth'].describe()
```

```python
# Calculate the age of customers at the time of enrollment as Age_Enrolled Column
df['Age_Enrolled'] = df['Dt_Customer'].dt.year - df['Year_Birth']
```

```python
df.loc[df['Age_Enrolled']>100,:]
```

```python
# Replace outliers in 'Age_Enrolled' with the median 'Year_Birth'
median_birth = int(df['Year_Birth'].median())
df.loc[df['Age_Enrolled'] > 100, 'Year_Birth'] = median_birth
df['Age_Enrolled'] = df['Dt_Customer'].dt.year - df['Year_Birth']
```

```python
df['Year_Birth'].min()
df['Age_Enrolled'].describe()
df.loc[df['ID'].isin([7829, 11004, 1150]), ['ID', 'Year_Birth', 'Age_Enrolled']]
```

```python
df['Marital_Status'].value_counts()
```

```python
# Filter the DataFrame to only include valid marital status values
valid_status = ['Married', 'Together', 'Single', 'Divorced', 'Widow']
df = df[df['Marital_Status'].isin(valid_status)].reset_index(drop=True)
df.shape
```

```python
df['Education'].value_counts()
```

```python
df['Education'] = df['Education'].replace({
    '2n Cycle': 'Masters',
    'Master': 'Masters',
    'Graduation': 'Bachelors'
})
df['Education'].value_counts()
```
